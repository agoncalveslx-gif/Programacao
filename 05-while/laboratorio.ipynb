## Atividade 1 · Inicializar, testar e atualizar

while testa a condição antes de executar o corpo. A atualização aproxima o programa de uma condição falsa.

**Primeiro:** prevê o resultado e executa o exemplo.

```python
n = 1
while n <= 3:
    print(n)
    n = n + 1
print('Valor final:', n)
```

## Atividade 3 · Pedir novamente um valor inválido

A primeira leitura ocorre antes do ciclo. Quando um valor é inválido, o corpo pede outro. Usa apenas entradas numéricas nesta atividade.

**Primeiro:** prevê o resultado e executa o exemplo.

```python
nota = float(input('Nota (0 a 20): '))
while nota < 0 or nota > 20:
    print('Fora do intervalo')
    nota = float(input('Nova nota: '))
print('Aceite:', nota)
```

### A tua experiência

```python
nota = float(input('Nota (0 a 20): '))
while nota < 0 or nota > 20:
    print('Fora do intervalo')
    nota = float(input('Nova nota: '))
print('Aceite:', nota)
```

## Atividade 4 · Terminar com uma sentinela

-1 assinala o fim; não faz parte dos dados. O zero é um valor válido e conta. Introduz 5, 0, 3 e -1.

**Primeiro:** prevê o resultado e executa o exemplo.

```python
total = 0
quantidade = 0
valor = int(input('Valor (-1 termina): '))
while valor != -1:
    total += valor
    quantidade += 1
    valor = int(input('Valor (-1 termina): '))
print('Total:', total)
print('Quantidade:', quantidade)
```
