## Atividade 2 · Contar ocorrências

get(palavra, 0) permite iniciar uma contagem sem aceder diretamente a uma chave ausente.

**Primeiro:** prevê o resultado e executa o exemplo.

```python
texto = 'ana rui ana eva rui ana'
frequencias = {}
for palavra in texto.split():
    frequencias[palavra] = frequencias.get(palavra, 0) + 1
for palavra, quantidade in frequencias.items():
    print(palavra, quantidade)
```

## Atividade 3 · Cruzar inscrições

Um set reúne elementos únicos. | faz a união, & a interseção e - a diferença. sorted produz uma lista ordenada para facilitar a leitura.

**Primeiro:** prevê o resultado e executa o exemplo.

```python
python = {101, 102, 103}
bd = {102, 104}
print('Ambas:', sorted(python & bd))
print('Só Python:', sorted(python - bd))
print('Pelo menos uma:', sorted(python | bd))
```

## Atividade 4 · Guardar dados para outra execução

pickle permite guardar objetos Python. Esta célula lê apenas o ficheiro que acabou de criar; não carregues ficheiros pickle de origem desconhecida, pois a leitura pode executar código.

**Primeiro:** prevê o resultado e executa o exemplo.

```python
import pickle

dados = {'alunos': {101: 'Ana', 102: 'Rui'},
         'inscritos': {101}}
with open('lab12_dados.pkl', 'wb') as f:
    pickle.dump(dados, f)
with open('lab12_dados.pkl', 'rb') as f:
    recuperados = pickle.load(f)
print(recuperados['alunos'])
print(sorted(recuperados['inscritos']))
```
