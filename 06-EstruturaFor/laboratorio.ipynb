## Atividade 6 · Encontrar o primeiro divisor

**Nível: Análise**

O código procura um divisor próprio de n, isto é, um inteiro de 2 a n-1 que divida n. Assume-se n >= 2.

### Previsão e execução

Prever o resultado e executar o exemplo. Quando exista um defeito, observar o comportamento antes de o corrigir.

```python
n = 15
encontrado = False
for divisor in range(2, n):
    if n % divisor == 0:
        print('Divisor:', divisor)
        encontrado = True
if not encontrado:
    print('Sem divisor próprio')
```
