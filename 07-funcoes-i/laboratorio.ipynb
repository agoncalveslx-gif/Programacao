## Atividade 4 · Uma função com decisão

return termina a chamada imediatamente. Uma função pode ter vários return, mas uma chamada só devolve um resultado.

**Primeiro:** prevê o resultado e executa o exemplo.

```python
def maior(a, b):
    if a >= b:
        return a
    return b

print(maior(8, 3))
print(maior(2, 9))
print(maior(5, 5))
```
