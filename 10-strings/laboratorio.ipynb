## Atividade 3 · Contar palavras e caracteres

split() sem argumento separa pelos espaços em branco e agrupa separadores consecutivos. count conta ocorrências de uma substring.

**Primeiro:** prevê o resultado e executa o exemplo.

```python
texto = 'Python é útil. Python é claro.'
palavras = texto.lower().split()
print(palavras)
print('Palavras:', len(palavras))
print('Ocorrências:', texto.lower().count('python'))
```
