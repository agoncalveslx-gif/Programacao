## Atividade 3 · O mesmo número, várias representações

`bin(numero)` mostra a representação binária com o prefixo 0b. `hex(numero)` usa hexadecimal com 0x. Estas ferramentas permitem confirmar conversões feitas no papel.

**Primeiro:** prevê o resultado e executa o exemplo.

```python
numero = 13
print('Decimal:', numero)
print('Binário:', bin(numero))
print('Hexadecimal:', hex(numero))
print('Verificação:', int('1101', 2))
```
