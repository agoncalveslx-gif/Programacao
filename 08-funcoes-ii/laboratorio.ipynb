## Atividade 4 · Testar com assert

assert verifica uma condição. Se ela for falsa, lança AssertionError na execução normal. Um teste que passa não mostra nada, por isso há um print final.

**Primeiro:** prevê o resultado e executa o exemplo.

```python
def media(a, b):
    return (a + b) / 2

assert media(10, 14) == 12
assert media(0, 0) == 0
print('Testes concluídos')
```
