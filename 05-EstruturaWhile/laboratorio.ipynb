## Atividade 4 · Terminar com uma sentinela

**Nível: Aplicação**

-1 assinala o fim; não faz parte dos dados. O zero é um valor válido e conta. Introduzir 5, 0, 3 e -1.

### Previsão e execução

Prever o resultado e executar o exemplo. Quando exista um defeito, observar o comportamento antes de o corrigir.

```python
total = 0
quantidade = 0
valor = int(input('Valor (-1 termina): '))
while valor != -1:
    total += valor
    quantidade += 1
    valor = int(input('Valor (-1 termina): '))
print('Total:', total)
print('Quantidade:', quantidade)
```
