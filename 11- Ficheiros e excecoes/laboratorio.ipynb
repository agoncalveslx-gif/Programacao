## Atividade 3 · Tratar uma conversão inválida

**Nível: Aplicação**

try tenta executar um bloco; except ValueError trata uma falha de conversão. finally executa nas duas situações apresentadas.

### Previsão e execução

Prever o resultado e executar o exemplo. Quando exista um defeito, observar o comportamento antes de o corrigir.

```python
texto = 'abc'
try:
    numero = int(texto)
except ValueError:
    print('Texto inválido')
else:
    print('Número:', numero)
finally:
    print('Tentativa concluída')
```

## Atividade 4 · Ignorar linhas inválidas e contar

**Nível: Aplicação**

O ficheiro de teste é criado pela célula, pelo que não é necessário carregar ficheiros externos. As linhas inválidas são contadas, sem entrarem na soma.

### Previsão e execução

Prever o resultado e executar o exemplo. Quando exista um defeito, observar o comportamento antes de o corrigir.

```python
with open('lab11_notas.txt', 'w', encoding='utf-8') as f:
    f.write('10\nabc\n14\n')
total = 0.0
validas = 0
invalidas = 0
with open('lab11_notas.txt', encoding='utf-8') as f:
    for linha in f:
        try:
            nota = float(linha)
        except ValueError:
            invalidas += 1
            continue
        total += nota
        validas += 1
print('Total:', total, 'Válidas:', validas, 'Inválidas:', invalidas)
```
