## Atividade 1 · O limite de range

range(início, fim, passo) inclui o início e exclui o fim. O for atribui cada valor à variável de controlo.

**Primeiro:** prevê o resultado e executa o exemplo.

```python
for n in range(1, 6):
    print(n)
```

## Atividade 2 · Somar apenas alguns valores

O for percorre os valores; o if seleciona os que contribuem para a soma e a contagem.

**Primeiro:** prevê o resultado e executa o exemplo.

```python
total = 0
quantidade = 0
for n in range(1, 11):
    if n % 2 == 0:
        total += n
        quantidade += 1
print('Total:', total, 'Quantidade:', quantidade)
```

## Atividade 3 · Interromper ou saltar uma iteração

break termina o ciclo. continue salta o resto da iteração atual, mas permite avançar para a seguinte.

**Primeiro:** prevê o resultado e executa o exemplo.

```python
for n in range(1, 7):
    if n == 3:
        break
    print(n)
```

## Atividade 4 · Percorrer uma grelha

O ciclo interior executa por completo para cada valor do ciclo exterior.

**Primeiro:** prevê o resultado e executa o exemplo.

```python
for linha in range(1, 3):
    for coluna in range(1, 4):
        print(linha, coluna)
```
