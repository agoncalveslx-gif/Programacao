## Atividade 2 · Dois nomes para a mesma lista

```python
a = [10, 20]
b = a
c = a.copy()
b[0] = 99
c.append(30)
print('a:', a)
print('b:', b)
print('c:', c)
```

## Atividade 3 · Selecionar e resumir

Uma compreensão constrói uma nova lista. sum soma os elementos e len conta-os.

**Primeiro:** prevê o resultado e executa o exemplo.

```python
notas = [8, 12, 15, 7, 18]
aprovados = [n for n in notas if n >= 10]
print('Aprovados:', aprovados)
print('Quantidade:', len(aprovados))
print('Média da turma:', sum(notas) / len(notas))
```

## Atividade 4 · Tuplos e representação gráfica

Um tuplo é ordenado e impede substituir os seus elementos. Um gráfico de barras permite comparar as notas. Matplotlib está disponível no ambiente Colab habitual.

**Primeiro:** prevê o resultado e executa o exemplo.

```python
import matplotlib.pyplot as plt

notas = [8, 12, 15, 7, 18]
resumo = (min(notas), max(notas))
print('Mínimo e máximo:', resumo)
plt.bar(range(1, len(notas) + 1), notas)
plt.xlabel('Aluno')
plt.ylabel('Nota')
plt.ylim(0, 20)
plt.show()
```
